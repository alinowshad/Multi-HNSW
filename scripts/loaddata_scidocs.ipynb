```python
single_pt_files = [
    '/data1/liuyaoyang/Papers/vectorDB/POQD-ICML25/Decompose_retrieval/output/ali/embs/msmarco/query_sv.pt',
]

mul_pt_files = [
    '/data1/liuyaoyang/Papers/vectorDB/POQD-ICML25/Decompose_retrieval/output/ali/embs/msmarco/query_mv.pt',
]
```

```python
# Read each .pt file
single_embeddings = []
count = 0
for pt_file in single_pt_files:
        count += 1
        if count % 1000 == 0:
                print(f"Processed {count} files")
        data = torch.load(pt_file)
        single_embeddings.extend(data)
```

```python
# Read each .pt file
muti_embeddings = []
count = 0
for pt_file in mul_pt_files:
        count += 1
        if count % 1000 == 0:
                print(f"Processed {count} files")
        data = torch.load(pt_file)
        muti_embeddings.extend(data)
```

```python
combined_embeddings = []
for i in range(len(single_embeddings)):
    data = torch.cat([muti_embeddings[i], single_embeddings[i].unsqueeze(0)], dim=0)
    combined_embeddings.append(data)
print(len(combined_embeddings))
```

```python
mul_embeddings = torch.cat([e for e in combined_embeddings], dim=0)
mul_embeddings = mul_embeddings.cpu().numpy()  # single, controlled copy
mul_embeddings.shape
```

```python
total_shape = (mul_embeddings.shape[0],) + mul_embeddings.shape[1:]

# Create memory-mapped output array
out = np.lib.format.open_memmap('/data/ali/msmarco-large-multi/full_multi_embeddings_msmarco-large_query.npy', mode='w+', dtype=mul_embeddings.dtype, shape=total_shape)

# Copy data sequentially
out[:mul_embeddings.shape[0]] = mul_embeddings
print(out.shape)
```

```python
single_embeddings_t = []
for i in range(len(single_embeddings)):
    single_embeddings_t.append(single_embeddings[i])
single_embeddings_t = np.array(single_embeddings_t)
print(single_embeddings_t.shape)
np.save(f'/data/ali/msmarco-large-single/full_single_embeddings_msmarco-large_query.npy', single_embeddings)
print("single_embeddings.npy saved")
```

```python
base_chunks = []
for i in range(len(combined_embeddings)):
    count = 0
    count += len(combined_embeddings[i])
    base_chunks.append(count)

base_chunks = np.array(base_chunks)
print(base_chunks.shape)
np.save(f'/data/ali/msmarco-large-multi/full_multi_chunk_num_msmarco-large_query.npy', base_chunks)
print("base_chunks.npy saved")
```

```python
from pathlib import Path
import re
import pandas as pd

from beir.retrieval.evaluation import EvaluateRetrieval

results_dir = Path('/home/ali/Decompose-Retrival/scidocs-complete-withoutmaxsimapprox')
# Accept either parents_k{K}_tk{QK}.csv or parents_k{K}_qk{QK}.csv
file_re = re.compile(r"^parents_k(\d+)_(?:t|q)k(\d+)\.csv$")

# Optional: attach QPS if we have a lookup table
qps_map = {}
qps_path = results_dir / 'scidocs-large-multi_hnsw-complete_k.csv'
if qps_path.exists():
    qps_df = pd.read_csv(qps_path)
    # file uses 'tk' but we treat it as 'qk'
    if 'tk' in qps_df.columns and 'qk' not in qps_df.columns:
        qps_df = qps_df.rename(columns={'tk': 'qk'})
    for _, r in qps_df.iterrows():
        try:
            qps_map[(int(r['k']), int(r['qk']))] = float(r['QPS'])
        except Exception:
            continue

rows = []

for fp in sorted(results_dir.glob('parents_k*.csv')):
    m = file_re.match(fp.name)
    if not m:
        continue

    k = int(m.group(1))
    qk = int(m.group(2))

    res_df = pd.read_csv(fp)
    required_cols = {'query_idx', 'parent_id', 'distance'}
    if not required_cols.issubset(set(res_df.columns)):
        raise ValueError(f"{fp.name} is missing columns: {sorted(required_cols - set(res_df.columns))}")

    # Build BEIR-style results: results[qid] = {doc_id: score}
    # Lower distance is better -> negate so higher score is better.
    results = {}
    for qid, group in res_df.groupby('query_idx'):
        group = group.sort_values('distance', ascending=True).head(k)
        results[str(int(qid))] = {
            str(int(pid)): -float(dist)
            for pid, dist in zip(group['parent_id'], group['distance'])
        }

    # Evaluate only on overlapping query IDs (avoids empty-score edge cases)
    common_qids = sorted(set(qrels.keys()) & set(results.keys()), key=lambda x: int(x))
    if len(common_qids) == 0:
        print(f"Skipping {fp.name}: no overlapping query IDs between qrels and results")
        continue

    qrels_common = {qid: qrels[qid] for qid in common_qids}
    results_common = {qid: results[qid] for qid in common_qids}

    model = EvaluateRetrieval()
    ndcg, _map, recall, precision = model.evaluate(qrels_common, results_common, k_values=[k])
    mrr = EvaluateRetrieval.evaluate_custom(qrels_common, results_common, k_values=[k], metric='mrr')

    rows.append({
        'run': fp.stem,
        'k': k,
        'qk': qk,
        'QPS': qps_map.get((k, qk), None),
        'MRR@k': mrr[f'MRR@{k}'],
        'NDCG@k': ndcg[f'NDCG@{k}'],
        'Recall@k': recall[f'Recall@{k}'],
    })

summary_df = pd.DataFrame(rows).sort_values(['k', 'qk']).reset_index(drop=True)
summary_path = results_dir / 'metrics_summary.csv'
summary_df.to_csv(summary_path, index=False)

print(f"Saved: {summary_path}")
summary_df
```

```python
import logging
import pytrec_eval

logger = logging.getLogger(__name__)

def evaluate(
        qrels: dict[str, dict[str, int]],
        results: dict[str, dict[str, float]],
        k_values: list[int],
        ignore_identical_ids: bool = True,
    ) -> tuple[dict[str, float], dict[str, float], dict[str, float], dict[str, float]]:
        if ignore_identical_ids:
            logger.info(
                "For evaluation, we ignore identical query and document ids (default), please explicitly set ``ignore_identical_ids=False`` to ignore this."
            )
            popped = []
            for qid, rels in results.items():
                for pid in list(rels):
                    if qid == pid:
                        results[qid].pop(pid)
                        popped.append(pid)

        ndcg = {}
        _map = {}
        recall = {}
        precision = {}

        for k in k_values:
            ndcg[f"NDCG@{k}"] = 0.0
            _map[f"MAP@{k}"] = 0.0
            recall[f"Recall@{k}"] = 0.0
            precision[f"P@{k}"] = 0.0

        map_string = "map_cut." + ",".join([str(k) for k in k_values])
        ndcg_string = "ndcg_cut." + ",".join([str(k) for k in k_values])
        recall_string = "recall." + ",".join([str(k) for k in k_values])
        precision_string = "P." + ",".join([str(k) for k in k_values])
        evaluator = pytrec_eval.RelevanceEvaluator(qrels, {map_string, ndcg_string, recall_string, precision_string})
        scores = evaluator.evaluate(results)

        for query_id in scores.keys():
            for k in k_values:
                ndcg[f"NDCG@{k}"] += scores[query_id]["ndcg_cut_" + str(k)]
                _map[f"MAP@{k}"] += scores[query_id]["map_cut_" + str(k)]
                recall[f"Recall@{k}"] += scores[query_id]["recall_" + str(k)]
                precision[f"P@{k}"] += scores[query_id]["P_" + str(k)]

        for k in k_values:
            ndcg[f"NDCG@{k}"] = round(ndcg[f"NDCG@{k}"] / len(scores), 5)
            _map[f"MAP@{k}"] = round(_map[f"MAP@{k}"] / len(scores), 5)
            recall[f"Recall@{k}"] = round(recall[f"Recall@{k}"] / len(scores), 5)
            precision[f"P@{k}"] = round(precision[f"P@{k}"] / len(scores), 5)

        for eval in [ndcg, recall]:  # here are the metrics

            for k in eval.keys():
                logger.info(f"{k}: {eval[k]:.4f}")

        return ndcg, recall  # ndcg, _map, recall, precision
```
